```python
dev.channels["A"].mode=pysmu.Mode.SVMI
dev.channels["B"].mode=pysmu.Mode.HI_Z
```

```python
s.start(0)
dev.channels["A"].write(sine_wave,-1)
```

```python
data = dev.read(300)
list1, list2, list3, list4 = [], [], [], []

# Loop through the items without unpacking them immediately
for item in data:
    # Check if the item is actually a tuple 
    if isinstance(item, tuple):
        # Now it is safe to unpack
        (a, b), (c, d) = item
        list1.append(a)
        list2.append(b)
        list3.append(c)
        list4.append(d)
    else:
        # If it hits the '...', it simply ignores it and moves on
        continue

print(list1)
print(list3)
print(len(list1))
print(len(list3))
```

```python
import numpy as np

x = np.asarray(list1, dtype=float)
y = np.asarray(list3[:len(x)], dtype=float)
# fs = sampling frequency

N = len(x)
fs = 10000

X = np.fft.fft(x)
Y = np.fft.fft(y)

freq = np.fft.fftfreq(N, d=1/fs)
```

```python
pos = freq > 0

peak_x = np.argmax(np.abs(X[pos]))
peak_y = np.argmax(np.abs(Y[pos]))

freq_x = freq[pos][peak_x]
freq_y = freq[pos][peak_y]

print("Input frequency =", freq_x)
print("Output frequency =", freq_y)
```
